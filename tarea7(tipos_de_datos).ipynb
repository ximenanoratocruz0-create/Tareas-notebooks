# Tarea 7. Tipos de datos de Phyton 3

## **int (enteros)**

```python
edad = 25
print(edad)
```

## **float (flotantes)**

```python
pi = 3.14
print(pi)
```

## **complex (complejos)**

```python
numero = 3 + 4j
print (numero)
```

## **str (cadena)**

```python
mensaje = "Aprendiendo phyton"
print(mensaje)
```

## **list (lista)**

```python
nums = [1,2,3,4]
print(nums)
print(nums[0])
print(nums[-1])
```

## **tuple (tuplas)**

```python
coords = (10, 20)
print(coords)
```

## **range (rango)**

```python
num =range(10)
print (num)
print (list(num))
```

## **dict (diccionario)**

```python
perfil = {"nombre": "Ximena", "profesion": "Ing. Industrial"}
print(perfil["nombre"])
print(perfil["profesion"])
```

## **set (conjunto)**

```python
frutas = {"manzana","platano"}
print(frutas)
```

## **frozenset (conjuntos congelados)**

```python
dias = frozenset({"lunes", "martes", "miercoles"})
print(dias)
```

## **bool (booleanos)**

```python
activo = True
print(activo)
```
